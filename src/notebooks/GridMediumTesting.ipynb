```julia
fpath = RayTracing.jmfp("/home/jmyslinski/random_stuff/PBRJ/test/test_grid-2x2x2.pbrt")
render_from_medium = RayTracing.Translate(RayTracing.Pnt3(0,0,0))
Le = RayTracing.spectrum_from_float(0.0)
sigma_a = RayTracing.spectrum_from_float(0.01)
sigma_s = RayTracing.spectrum_from_float(0.015)
sigma_scale = 1.0
p0 = RayTracing.Pnt3(0,0,0)
p1 = RayTracing.Pnt3(1,1,1)
g = 0.0
majorant_grid_res = RayTracing.Pnt3(5, 5, 5)
```

```julia
grid_medium = RayTracing.GridMedium(
    fpath,
    render_from_medium,
    sigma_a,
    sigma_s,
    sigma_scale,
    p0,
    p1,
    g,
    majorant_grid_res
)
```
