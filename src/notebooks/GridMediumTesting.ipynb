```julia
include("../RayTracing.jl")
```

```julia
fpath = RayTracing.jmfp("/Users/johnmyslinski/Documents/pbrt-v3-scenes/cloud/geometry/density_render.70.pbrt")
render_from_medium = RayTracing.Translate(RayTracing.Pnt3(-1.0, 0.0, -1.2))
Le = RayTracing.spectrum_from_float(0.0)
sigma_a = RayTracing.spectrum_from_float(0.01)
sigma_s = RayTracing.spectrum_from_float(0.015)
sigma_scale = 1.0
p0 = RayTracing.Pnt3(0.01, 0.01, 0.01)
p1 = RayTracing.Pnt3(1.99, 1.99, 0.79)
g = 0.0
majorant_grid_res = RayTracing.Pnt3(16, 16, 16)



sphere_transform = RayTracing.Translate(RayTracing.Pnt3(.045, 1.045, -.75))
sphere = RayTracing.Sphere(
    RayTracing.ShapeCore(
        sphere_transform,
        RayTracing.Inv(sphere_transform),
        false,
        false
    ),
    1.4
)
```

```julia
grid_medium = RayTracing.GridMedium(
    fpath,
    render_from_medium,
    sigma_a,
    sigma_s,
    sigma_scale,
    p0,
    p1,
    g,
    majorant_grid_res
)
```

```julia
RayTracing.max_value(
    grid_medium.density_grid, 
    RayTracing.Bounds3(
        RayTracing.Pnt3(0,0,0),
        RayTracing.Pnt3(0.2,0.2,0.2),
    ))
```
