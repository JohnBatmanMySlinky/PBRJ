```julia
tb_min = RayTracing.Pnt2(1,1)
tb_max = RayTracing.Pnt2(2,2)

tile_bounds = RayTracing.Bounds2(tb_min, tb_max)

for pixel in tile_bounds
    println(pixel)
end
```
