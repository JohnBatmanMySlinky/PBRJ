```julia
include("../RayTracing.jl")
```

```julia
orb_translate = RayTracing.RotateZ(-135.0)
orb =  RayTracing.parse_obj(
    "../../ref/dragon1.obj",
    orb_translate,
    false,
    false,
    nothing
)
print("OK")
```

# the center

```julia
RayTracing.mean(RayTracing.mean.([RayTracing.get_vertices(orb[i]) for i in 1:length(orb)]))
```

# min

```julia
RayTracing.minimum(RayTracing.minimum.([RayTracing.get_vertices(orb[i]) for i in 1:length(orb)]))
```

# max

```julia
RayTracing.maximum(RayTracing.maximum.([RayTracing.get_vertices(orb[i]) for i in 1:length(orb)]))
```
