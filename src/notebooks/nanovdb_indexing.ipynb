```julia
struct GDM
    nx::Int64
    ny::Int64
    nz::Int64
    density::Vector{Float64}

    function GDM(n, density)
        @assert n * n * n == length(density)
        return new(n, n, n, density)
    end
end

function lerp(t::Float64, a::Float64, b::Float64)::Float64
    return a + t * (b - a)
end

function D(gdm::GDM, p::RayTracing.Pnt3)::Float64
    sample_bounds = RayTracing.Bounds3(RayTracing.Pnt3(0,0,0), RayTracing.Pnt3(gdm.nx, gdm.ny, gdm.nz))

    print("\tp: $(p)\n")

    if !RayTracing.inside_exclusive(p, sample_bounds)
        return 0.0
    else
        print("\ti: $(Int((p.z * gdm.ny + p.y) * gdm.nx + p.x + 1))\n")
        return gdm.density[Int((p.z * gdm.ny + p.y) * gdm.nx + p.x + 1)]
    end
end

function density(gdm::GDM, p::RayTracing.Pnt3)::Float64
    psample = RayTracing.Pnt3(p.x * gdm.nx - 0.5, p.y * gdm.ny - 0.5, p.z * gdm.nz - 0.5)
    pfloor = floor.(psample)
    d = psample - pfloor

    print("psample: $(psample)\n")
    print("pfloor: $(pfloor)\n")

    d00 = RayTracing.lerp(d.x, D(gdm, pfloor),               D(gdm, pfloor + RayTracing.Pnt3(1,0,0)))
    d10 = RayTracing.lerp(d.x, D(gdm, pfloor + RayTracing.Pnt3(0,1,0)), D(gdm, pfloor + RayTracing.Pnt3(1,1,0)))
    d01 = RayTracing.lerp(d.x, D(gdm, pfloor + RayTracing.Pnt3(0,0,1)), D(gdm, pfloor + RayTracing.Pnt3(1,0,1)))
    d11 = RayTracing.lerp(d.x, D(gdm, pfloor + RayTracing.Pnt3(0,1,1)), D(gdm, pfloor + RayTracing.Pnt3(1,1,1)))
    d0  = RayTracing.lerp(d.y, d00, d10)
    d1  = RayTracing.lerp(d.y, d01, d11)
    return RayTracing.lerp(d.z, d0, d1)
end
```
