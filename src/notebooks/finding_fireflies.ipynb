```julia
using OpenEXR
```

# Version A

```julia
img = OpenEXR.load("../../scratch/finding_fireflies/a_pbrj.exr")
```

```julia
img2 = OpenEXR.load("../../scratch/finding_fireflies/a_pbrt.exr")
```

```julia
diff = copy(img)
for x in 1:250
    for y in 1:250
        diff[x,y] = img[x,y].r - img2[x,y].r
    end
end
diff
```

```julia
diff = copy(img)
for x in 1:250
    for y in 1:250
        diff[x,y] = img[x,y].g - img2[x,y].g
    end
end
diff
```

```julia
diff = copy(img)
for x in 1:250
    for y in 1:250
        diff[x,y] = img[x,y].b - img2[x,y].b
    end
end
diff
```
