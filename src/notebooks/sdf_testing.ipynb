```julia
ray = RayTracing.Ray(
    RayTracing.Pnt3(5, 5, 5),
    RayTracing.Vec3(.1, .1, .1),
    0.0,
    typemax(Float64)
)

s = RayTracing.SDF_Sphere(
    RayTracing.ShapeCore(),
    1.0
)
RayTracing.intersect(s, ray)
```
