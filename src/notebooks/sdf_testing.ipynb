```julia
struct Node
    contents::Union{Function, Number}
    left::RayTracing.Maybe{Node}
    right::RayTracing.Maybe{Node}

    Node(contents) = new(contents, nothing, nothing)

    Node(contents, left, right) = new(contents, left, right)
end
```

```julia
function f(node::Node)
    if node.contents isa Number
        return node.contents
    elseif node.contents isa Function
        @assert !((node.left isa Nothing) && (node.right isa Nothing))

        return node.contents(f(node.left), f(node.right))
    else
        @assert false
    end
end

add(a, b) = a + b
subtract(a, b) = a - b
multiply(a, b) = a * b
divide(a, b) = a / b
power(a, b) = a ^ b
```

```julia
# (3+5) * (10-2) = 64
left_subtree = Node(add, Node(3), Node(5))
right_subtree = Node(subtract, Node(10), Node(2))
tree = Node(multiply, left_subtree, right_subtree)

f(tree)
```

```julia
# (4^2) / (6+3) = 1.777777
tree2 = Node(
    divide,
    Node(power, Node(4), Node(2)),
    Node(add, Node(6), Node(3))
)
f(tree2)
```

```julia
# wrapper around SDF Functions. Holds the function and K smoothing parameter
struct SDFFunction
    k::Float64
    F::Function
end

# The Tree
struct SDFNode <: RayTracing.ImplicitSurface
    contents::Union{SDFFunction, RayTracing.ImplicitSurface}
    core::RayTracing.Maybe{RayTracing.ShapeCore}
    bounding_sphere::RayTracing.Maybe{RayTracing.Sphere}
    left::RayTracing.Maybe{SDFNode}
    right::RayTracing.Maybe{SDFNode}

    # Function & SDF Node
    SDFNode(contents) = new(contents, nothing, nothing, nothing)

    # Tree
    SDFNode(contents, core, bs, left, right) = new(contents, core, bs, left, right)
end
```

```julia
function RayTracing.f(node::SDFNode, t::Float64, r::RayTracing.AbstractRay)::Float64
    return f(node, RayTracing.at(r,t))
end

function RayTracing.f(node::SDFNode, pp::RayTracing.Pnt3)::Float64
    if node.contents isa RayTracing.ImplicitSurface
        return RayTracing.f(node.contents, node.core.world_to_object(pp))
    elseif node.contents isa SDFFunction
        @assert !((node.left isa Nothing) && (node.right isa Nothing))

        l = f(node.left, node.left.contents.core.world_to_object(pp))
        r = f(node.right, node.right.contents.core.world_to_object(pp))
        return node.contents.F(l, r, node.contents.k)
    else
        @assert false
    end
end

function RayTracing.f(s::RayTracing.ImplicitSurface, t::Float64, r::RayTracing.AbstractRay)
    pp = RayTracing.at(r,t)
    return f(s,pp)
end
```

```julia
tmp = RayTracing.Pnt3(0.0, 0.0, 0.0)
s1_sc = RayTracing.ShapeCore(
    RayTracing.Translate(tmp),
    RayTracing.Inv(RayTracing.Translate(tmp)),
    false,
    false
)
s1 = RayTracing.SDF_Box(
    s1_sc,
    RayTracing.Vec3(1,1,1)
)

tmp = RayTracing.Pnt3(0.0, 0.5, 0.0)
s2_sc = RayTracing.ShapeCore(
    RayTracing.Translate(tmp),
    RayTracing.Inv(RayTracing.Translate(tmp)),
    false,
    false
)
s2 = RayTracing.SDF_Sphere(
    s2_sc,
    0.5
)

func = SDFFunction(
    0.0,
    RayTracing.sdf_union
)

bs = RayTracing.Sphere(
    RayTracing.Pnt3(0,0,0),
    2.0
)

sc = RayTracing.ShapeCore()

tree = SDFNode(func, sc, bs, SDFNode(s1), SDFNode(s2))

@assert f(tree, si.core.p) == 0.0
```

```julia
ray = RayTracing.Ray(
    RayTracing.Pnt3(5, 5, 5),
    RayTracing.Vec3(-.1, -.1, -.1),
    0.0,
    typemax(Float64)
)

RayTracing.intersect(tree, ray)
```
