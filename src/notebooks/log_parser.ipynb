```julia
function findall(pattern, str)
    to_return = UnitRange{Int64}[]
    s = 1
    while true
        range = findnext(pattern, str, s)
        if range == nothing
            break
        else
            push!(to_return, range)
            s = first(range)+1
        end
    end
    return to_return
end

function print_buffed(str, ranges, lbuff, rbuff)
    for (a,b) in ranges
        aa = max(1, a-lbuff)
        bb = min(length(str), b+rbuff)
        print(str[aa:bb],"\n")
    end
end
```

```julia
cells_to_check = [58.5, 59.5, 60.5]
for x in cells_to_check
    for y in cells_to_check
        print_buffed(jlog, findall("$(x), $(y)], ", jlog), 20, 110)
        print_buffed(clog, findall(" $(x), $(y) ], ", clog), 20, 70)
        print("\n")
    end
end
```
