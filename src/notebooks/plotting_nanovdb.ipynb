```python
import matplotlib.pyplot as plt
import numpy as np
from dataclasses import dataclass
```

```python
@dataclass
class Pnt3:
    x: float
    y: float
    z: float
```

```python
circle_translate = Pnt3(-9.984, 73.008, -42.64)
circle_scale = Pnt3(206.544, 140.4, 254.592)
bbox_pmin = Pnt3(-217.5, -67.5, -298.33331298828125)
bbox_pmax = Pnt3(197.5, 214.16665649414062, 212.5)
```

```python
# Define the radius
radius = 1.44224957031

# Generate points for the circle
theta = np.linspace(0, 2 * np.pi, 100)
# x = (radius * np.cos(theta) + circle_translate.x) * circle_scale.x
# z = (radius * np.sin(theta) + circle_translate.z) * circle_scale.z

x = (radius * np.cos(theta) * circle_scale.x + circle_translate.x) 
z = (radius * np.sin(theta) * circle_scale.z + circle_translate.z)

corners_x = [bbox_pmin.x, bbox_pmax.x, bbox_pmax.x, bbox_pmin.x, bbox_pmin.x]
corners_y = [bbox_pmin.y, bbox_pmin.y, bbox_pmax.y, bbox_pmax.y, bbox_pmin.y]

# Plot the circle
plt.figure(figsize=(6,6))
plt.plot(x, z, 'r-')

# plot the square
plt.plot(corners_x, corners_y, 'b-')
plt.fill(corners_x, corners_y, 'lightblue')

# options
plt.gca().set_aspect('equal', adjustable='box')
plt.title(f'The World')
plt.xlabel('X')
plt.ylabel('Z')
plt.grid(True)
plt.show()
```
