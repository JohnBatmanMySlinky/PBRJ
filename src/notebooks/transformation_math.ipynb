```julia
include("../RayTracing.jl")
```

```julia
origin = RayTracing.Pnt3(-5, -5, -5)
target = RayTracing.Pnt3(1.2, 3.4, 4.9)
bounds = RayTracing.Bounds3(
    RayTracing.Pnt3(0, 0, 0),
    RayTracing.Pnt3(7, 7, 7)
)
@assert RayTracing.inside(target, bounds)
@assert !RayTracing.inside(origin, bounds)

ray = RayTracing.Ray(
    origin,
    RayTracing.normalize(RayTracing.Vec3(target - origin)),
    0.0,
    typemax(Float64)
)

tr1 = RayTracing.Scale(RayTracing.Vec3(2, 2, 2))
tr2 = RayTracing.Scale(RayTracing.Vec3(4, 2.3, 3.9))
```

# Baseline

```julia
check, tmin, tmax = RayTracing.intersect_p(bounds, ray)
print("Hit Results: \n\t$(check), $(tmin), $(tmax)\n\t$(RayTracing.at(ray, tmin))\n\t$(RayTracing.at(ray, tmax))")
```

# Uniform Scaling

```julia
check, tmin, tmax = RayTracing.intersect_p(tr1(bounds), tr1(ray))
print("Hit Results: \n\t$(check), $(tmin), $(tmax)\n\t$(RayTracing.at(tr1(ray), tmin))\n\t$(RayTracing.at(tr1(ray), tmax))\n")
print("Aka:\n\t$(RayTracing.at(ray, tmin))\n\t$(RayTracing.at(ray, tmax))")
```

# Non-Uniform Scaling

```julia
check, tmin, tmax = RayTracing.intersect_p(tr2(bounds), tr2(ray))
print("Hit Results: \n\t$(check), $(tmin), $(tmax)\n\t$(RayTracing.at(tr2(ray), tmin))\n\t$(RayTracing.at(tr2(ray), tmax))\n")
print("Aka:\n\t$(RayTracing.at(ray, tmin))\n\t$(RayTracing.at(ray, tmax))")
```
