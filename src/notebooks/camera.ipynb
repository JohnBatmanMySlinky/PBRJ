```julia
include("../RayTracing.jl")
```

```julia
filter = RayTracing.BoxFilter(RayTracing.Pnt2(.5, .5))

# Instantiate a Film
film = RayTracing.Film(
    RayTracing.Pnt2(360, 360),
    RayTracing.Bounds2(RayTracing.Pnt2(0, 0), RayTracing.Pnt2(1, 1)),
    filter,
    1.0,
    1.0,
    "yeehaw.exr"
)

# Instantiate a Camera
look_from = RayTracing.Pnt3(648.064, -82.473, -63.856)
look_at = RayTracing.Pnt3(6.021, 100.043, -43.679)
up = RayTracing.Vec3(0.273, 0.962, -0.009)
screen = RayTracing.Bounds2(RayTracing.Pnt2(-1, -1), RayTracing.Pnt2(1, 1))
C = RayTracing.PerspectiveCamera(RayTracing.LookAt(look_from, look_at, up), screen, 0.0, 1.0, 0.0, 1e6, 31.07, film)
```

```julia
cs = RayTracing.CameraSample(
    RayTracing.Pnt2(0.5, 0.5),
    RayTracing.Pnt2(0.5, 0.5),
    0.5
)
```

```julia
ray, _ = RayTracing.generate_ray_differential(C, cs)
```
