```julia
function normal(meta_balls::RayTracing.MetaBalls, pp::RayTracing.Pnt3)::RayTracing.Vec3
    n = RayTracing.Nml3(0, 0, 0)
    for s in meta_balls.ks
        p = RayTracing.norm(pp-s)
        if p <= 3.0
            x =  0.4444 * 6.0 * (pp.x - s.x) * p^4 / 3.0 ^ 6
                -1.8888 * 4.0 * (pp.x - s.x) * p^2 / 3.0 ^ 4
                 2.4444 * 2.0 * (pp.x - s.x)       / 3.0 ^ 2

            y =  0.4444 * 6.0 * (pp.y - s.y) * p^4 / 3.0 ^ 6
                -1.8888 * 4.0 * (pp.y - s.y) * p^2 / 3.0 ^ 4
                 2.4444 * 2.0 * (pp.y - s.y)       / 3.0 ^ 2

            z =  0.4444 * 6.0 * (pp.z - s.z) * p^4 / 3.0 ^ 6
                -1.8888 * 4.0 * (pp.z - s.z) * p^2 / 3.0 ^ 4
                 2.4444 * 2.0 * (pp.z - s.z)       / 3.0 ^ 2

            n += RayTracing.Nml3(x,y,z)
        end
    end
    return RayTracing.normalize(n)
end
```
