```julia
include("../RayTracing.jl")
using Roots
```

```julia
parsed_args = RayTracing.parse_commandline()
```

```julia
parsed_args["scene-number"] = 5
```

```julia
I, scene = RayTracing.build_scene(parsed_args)
```

```julia
ray = RayTracing.Ray(
    RayTracing.Pnt3(30, 30, 30),
    RayTracing.normalize(RayTracing.Pnt3(0, 0, 0)-RayTracing.Pnt3(30, 30, 30)),
    0.0,
    typemax(Float64)
)

check, t, si = RayTracing.intersect!(scene.b, ray)
if !(si isa Nothing)
    si.shape.core.world_to_object(si.core.p)
end
```
