```julia
include("../RayTracing.jl")
```

```julia
fpath = RayTracing.jmfp("/Users/johnmyslinski/Documents/pbrt-v4-volumes/scenes/anemone/geometry/anemone_medium.pbrt")
nx, ny, nz, d, le_grid, temperature_grid, p0, p1 = RayTracing.parse_media(fpath)
```

```julia
function downsample_3d_systematic(array, X, Y, Z, NTH)
    SG = RayTracing.SampledGrid(array, X, Y, Z)
    XX = Int(floor(X * NTH))
    YY = Int(floor(Y * NTH))
    ZZ = Int(floor(Z * NTH))

    downsampled = Float64[]
    for x in 1:XX
        for y in 1:YY
            for z in 1:ZZ
                push!(downsampled, RayTracing.lookup(SG, RayTracing.Pnt3(x/XX, y/YY, z/ZZ)))
            end
        end
    end
    return downsampled, (XX, YY, ZZ)
end
```

```julia
downsampled_le, (nx2, ny2, nz2) = downsample_3d_systematic(le_grid, nx, ny, nz, .3)
```

```julia
downsampled_d, (nx2, ny2, nz2) = downsample_3d_systematic(d, nx, ny, nz, .3)
```

```julia
@assert length(downsampled_le) == length(downsampled_d)
```

```julia
@assert length(downsampled_d) == nx2 * ny2 * nz2
```

```julia
new_pbrt_file_1 = "MakeNamedMedium \"/obj/medium_anemone/timeshift1[1,0]\" \"string type\" \"uniformgrid\" \n\"rgb Le\" [ 10 0.5 5 ] \n\"float Lescale\" [ "
new_pbrt_file_2 = " ] \n\"float density\" [ "
new_pbrt_file_3 = " ] \n\"integer nx\" [ "
new_pbrt_file_4 = " ] \n\"integer ny\" [ "
new_pbrt_file_5 = " ] \n\"integer nz\" [ "
new_pbrt_file_6 = " ] \n\"point3 p0\" [ "
new_pbrt_file_7 = " ] \n\"point3 p1\" [ "
new_pbrt_file_8 = " ] \n\"rgb sigma_a\" [ 0.1 0.9 0.5 ] \n\"rgb sigma_s\" [ 0.2 0.01 1 ]"
```

```julia
open("/Users/johnmyslinski/Documents/pbrt-v4-volumes/scenes/anemone/geometry/anemone_medium_downsampled.pbrt", "w") do io
    write(io, new_pbrt_file_1)
    write(io, join(downsampled_le, " "))
    write(io, new_pbrt_file_2)
    write(io, join(downsampled_d, " "))
    write(io, new_pbrt_file_3)
    write(io, join(nx2, " "))
    write(io, new_pbrt_file_4)
    write(io, join(ny2, " "))
    write(io, new_pbrt_file_5)
    write(io, join(nz2, " "))
    write(io, new_pbrt_file_6)
    write(io, join(p0, " "))
    write(io, new_pbrt_file_7)
    write(io, join(p1, " "))
    write(io, new_pbrt_file_8)
end
```
