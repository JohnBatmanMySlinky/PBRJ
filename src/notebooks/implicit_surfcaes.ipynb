```julia
include("../RayTracing.jl")
using Roots
```

```julia
ray = RayTracing.Ray(RayTracing.Pnt3(10, 10, 10), RayTracing.Vec3(-1, -1, -1), 0.0, 999999999.0)
ks = [RayTracing.Pnt3(0,0,0)]
ts = [9.13397, 10.866]
```

```julia
for t in ts
    magic = 0.5 # what we want the solution to be 
    f = 0.0 - magic # subtract from f so we can "solve for zero"
    R = 3.0 
    for k in ks
        r = RayTracing.norm(RayTracing.at(ray, t)-k)
        if r <= R
            f += -.4444 * (r ^ 6) / (R ^ 6) + 1.8888 * (r ^ 4) / (R ^ 4) - 2.4444 * (r ^ 2) / (R ^ 2) + 1.0
        end
    end
    print("t: $(t), f: $(f)\n\n")
end
```
