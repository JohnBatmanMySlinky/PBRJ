```julia
struct Spectrum
    x::Float64
    y::Float64
    z::Float64
end

# Define the abstract type with type parameter
abstract type Texture{T <: Union{Float64, Spectrum}} end

# Make ConstantTexture a subtype of Texture
struct ConstantTexture{T <: Union{Float64, Spectrum}} <: Texture{T}
    value::T
end

struct MatteMaterial
    color::Texture{Spectrum}
    roughness::Texture{Float64}
end

# The evaluate function can now be defined for the abstract type
function evaluate(t::Texture{T})::T where T <: Union{Spectrum, Float64}
    error("evaluate not implemented for $(typeof(t))")
end

# Implementation for ConstantTexture
function evaluate(t::ConstantTexture{T})::T where T <: Union{Spectrum, Float64}
    return t.value
end
```

```julia
ts = ConstantTexture(Spectrum(1,2,3))
tf = ConstantTexture(.01)


matte = MatteMaterial(ts, tf)
```
