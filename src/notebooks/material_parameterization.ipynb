```julia
struct Spectrum
    x::Float64
    y::Float64
    z::Float64
end

struct ConstantTexture{T <: Union{Float64, Spectrum}}
    value::T
end

struct ConstantTexture2
    value::Union{Spectrum, Float64}
end

struct MatteMaterial
    color::ConstantTexture{Spectrum}
    roughness::ConstantTexture{Float64}
end

struct MatteMaterial2
    color::ConstantTexture2
    roughness::ConstantTexture2
end

function evaluate(t::ConstantTexture{T})::T where T <: Union{Spectrum, Float64}
    return t.value
end

function evaluate(t::ConstantTexture)
    return t.value
end
```

```julia
ts = ConstantTexture(Spectrum(1,2,3))
tf = ConstantTexture(.01)

ts2 = ConstantTexture2(Spectrum(1,2,3))
tf2 = ConstantTexture2(.01)

matte = MatteMaterial(ts, tf)
matte2 = MatteMaterial2(ts2, tf2)
```
