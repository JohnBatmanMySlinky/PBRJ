```julia
struct Pnt3
    x::Float64
    y::Float64
    z::Float64
end

mutable struct Sphere
    center::Pnt3
    radius::Float64
end
```

```julia
function SampleT_maj(callback, x)
    print(sphere)
    print(x)
    if callback(sphere)
        return 0.0
    else
        return x^2
    end
end

function render()
    new_is = Float64[]
    spheres = Sphere[]
    for i in 1:10
        i = Float64(i)
        sphere = Sphere(Pnt3(i, i, i), i)
        new_i = SampleT_maj(i) do sphere
            return mod(sphere.radius, 2) == 0
        end
        push!(new_is, new_i)
        push!(spheres, sphere)
    end
    return new_is
end
```

```julia
render()
```

```julia
# Real-world example with state modification
function process_with_state(callback::Function, x::Int)
    state = Dict("count" => 0)
    log = []
    for i in 1:10
        check = callback(state, i)
        if check
            push!(log, (x, i))
        end
    end
    return (state, log)
end

# Using do block with state modification
final_state = process_with_state(5) do state, i
    state["count"] += i
    return mod(i,2) == 0
end
```
