```julia
file_path = "/Users/johnmyslinski/Documents/PBRJ/ref/floating_lanterns/autobackup02_3.obj"
trans = RayTracing.Translate(RayTracing.Pnt3(0, 0, 0))
```

```julia
for (i, (k, v)) in enumerate(sort(collect(group_to_idx), by=x->x[2]))
    println("$i - $v - $k")
end
```

```julia
function HELP(idx)
    prim = RayTracing.Primitive[]
    for tri in tris[idx]
        push!(prim, RayTracing.Primitive(tri, "dummy", nothing))
    end
    bvh = RayTracing.BVH(prim)
    wb = RayTracing.world_bounds(bvh)
    
    print("""
        Centroid: $(RayTracing.centroid(wb))
            pMin: $(wb.pMin)
            pMax: $(wb.pMax)
    """)
    return nothing
end
```

```julia
HELP(4) # water
```

```julia
HELP(24) # ground
```

```julia
HELP(3) # wall
```
