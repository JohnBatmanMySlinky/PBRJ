```julia
function HELP(idx, tris)
    prim = RayTracing.Primitive[]
    for tri in tris[idx]
        push!(prim, RayTracing.Primitive(tri, "dummy", nothing))
    end
    bvh = RayTracing.BVH(prim)
    wb = RayTracing.world_bounds(bvh)
    
    print("""
        Centroid: $(RayTracing.centroid(wb))
            pMin: $(wb.pMin)
            pMax: $(wb.pMax)
    """)
    return nothing
end
```

```julia
function write_obj(filename::String, tris)
    open(filename, "w") do io
        v_offset = 0
        vt_offset = 0
        vn_offset = 0
        
        for tri in tris
            # Write vertices
            for v in tri.vertices
                println(io, "v $(v[1]) $(v[2]) $(v[3])")
            end
            
            # Write texture coordinates
            for uv in tri.uvs
                println(io, "vt $(uv[1]) $(uv[2])")
            end
            
            # Write normals
            for n in tri.normals
                println(io, "vn $(n[1]) $(n[2]) $(n[3])")
            end
            
            # Write face (1-indexed)
            i1, i2, i3 = v_offset + 1, v_offset + 2, v_offset + 3
            println(io, "f $i1/$i1/$i1 $i2/$i2/$i2 $i3/$i3/$i3")
            
            v_offset += 3
            vt_offset += 3
            vn_offset += 3
        end
    end
end
```

```julia
# Usage:
lamp_dump = "/Users/johnmyslinski/Documents/PBRJ/ref/floating_lanterns/lamp_extraction.obj"
write_obj(lamp_dump, tris[29])
```
