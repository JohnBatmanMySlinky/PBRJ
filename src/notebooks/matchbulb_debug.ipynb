```julia
paths = [
    "/Users/johnmyslinski/Documents/pbrt-v4-volumes/scenes/matchbulb/geometry/bulb_contact_ascii.obj",
    "/Users/johnmyslinski/Documents/pbrt-v4-volumes/scenes/matchbulb/geometry/ceramic_ascii.obj",
    "/Users/johnmyslinski/Documents/pbrt-v4-volumes/scenes/matchbulb/geometry/glass_bulb_ascii.obj",
    "/Users/johnmyslinski/Documents/pbrt-v4-volumes/scenes/matchbulb/geometry/bulb_cap_ascii.obj",
    "/Users/johnmyslinski/Documents/pbrt-v4-volumes/scenes/matchbulb/geometry/matchstick_ascii.obj",
    "/Users/johnmyslinski/Documents/pbrt-v4-volumes/scenes/matchbulb/geometry/ground_ascii.obj"
]
for path in paths

    primitives = RayTracing.Primitive[]
    bulb_contact = RayTracing.parse_obj(
        RayTracing.jmfp(path), 
        RayTracing.Translate(RayTracing.Pnt3(0,0,0)),
        false,
        false,
        nothing
    )
    for tris in bulb_contact
        for tri in tris
            push!(primitives, RayTracing.Primitive(tri, "mat_aluminum", nothing))
        end
    end

    bvh = RayTracing.BVH(primitives)

    wb = RayTracing.world_bounds(bvh)
    c = RayTracing.centroid(wb)
    println(split(path,'/')[end] * "\nWB: $wb\nC: $c\n")
end
```
