```julia
using Images
using Random

# Set a specific seed for reproducibility
function create_colorful_grid(seed, X, Y)
    # Set the random seed
    Random.seed!(seed)
    
    # Initialize a 10x10 RGB image
    img = zeros(RGB{N0f8}, X, Y)
    
    # Fill each pixel with a different random color
    for i in 1:X
        for j in 1:Y
            # Generate random RGB values (0-1 for Julia's RGB type)
            r = rand()
            g = rand()
            b = rand()
            
            # Set pixel color
            img[i, j] = RGB(r, g, b)
        end
    end
    
    # Save the image
    save("colorful_grid_seed_$(seed)_$(X)x$(Y).png", img)
    
    return img
end

# Create and display the image with a specific seed
colorful_grid = create_colorful_grid(42, 5, 9)  # Using seed value 42

# To create another image with the same pattern, use the same seed
# colorful_grid_same = create_colorful_grid(42)  # Will be identical

# To create a different pattern, use a different seed
# colorful_grid_different = create_colorful_grid(100)  # Different pattern
```
