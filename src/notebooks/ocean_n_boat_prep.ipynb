# sampling functions

```julia
plane = obj[1]
weights = zeros(Float64, length(plane))
for (i, tri) in enumerate(plane)
    weights[i] = RayTracing.area(tri)
end

tri_sampling_dist = RayTracing.Distribution1D(weights)
letter_sampling_dist = RayTracing.Distribution1D(ones(Float64, 26))

N_LETTERS = 100

for n in 1:N_LETTERS
    chose_letter = rand()
    chose_tri = rand()
    chose_p = RayTracing.Pnt2(rand(), rand())

    letter_idx, _, _ = RayTracing.sample_discrete(letter_sampling_dist, chose_letter)

    tri_idx, _, _ = RayTracing.sample_discrete(tri_sampling_dist, chose_tri)
    p, n, uv, _ = RayTracing.sample(plane[tri_idx], chose_p)
end
```
