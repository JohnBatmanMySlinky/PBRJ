```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
N = 25
ks = [(N//2, N//2), (N//1.25, N//1.25)]
R = 3.0

def norm(a, b):
    return np.sqrt((a[0]-b[0])**2 + (a[1]-b[1])**2)

def f_inner(r, R):
    return -.4444 * (r ** 6) / (R ** 6) + 1.8888 * (r ** 4) / (R ** 4) - 2.4444 * (r ** 2) / (R ** 2) + 1.0

def f(ks, R, p):
    f_val = 0
    for k in ks:
        r = norm(k, p)
        if r <= R:
            f_val += f_inner(r, R)
    return f_val * 100

def plot(N, ks, R):
    grid = np.zeros((N, N))
    for x in range(N):
        for y in range(N):
            grid[x,y] += f(ks, R, (x,y))
    return grid

grid = plot(N, ks, R)

plt.imshow(grid)
plt.show()
```
