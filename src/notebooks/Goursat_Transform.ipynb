```julia
include("../RayTracing.jl")
# logger = RayTracing.NullLogger()
logger = RayTracing.SimpleLogger()       
RayTracing.global_logger(logger)
```

# the identity case

```julia
goursat_t = RayTracing.Translate(RayTracing.Pnt3(0,0,0))
goursat_sc = RayTracing.ShapeCore(
    goursat_t,
    RayTracing.Inv(goursat_t),
    false,
    false
)
goursat = RayTracing.GoursatSurface(
    goursat_sc,
    0.0,
    -3.0,
    -1.0
)

r = RayTracing.Ray(RayTracing.Pnt3(5, 5, 5), RayTracing.Vec3(-1, -1, -1), 0, typemax(Float64))

check, t, intr = RayTracing.intersect(goursat, r)
print("Final intersection: $(intr.core.p)\n")
```
