# let's assert that within an object in an obj, it is always either tris or quads..... no mixing please

```julia
objs = String[]
dirpath = "/Users/johnmyslinski/Documents/pbrt-v3-scenes/barcelona-pavilion/geometry/"
for (root, dir, files) in walkdir(dirpath)
    for file in files
        if endswith(file, ".obj")
            push!(objs, file)
        end
    end
end
```

```julia
for obj in objs
    file_path = dirpath * obj

    f_tested = 0 
    o_tested = 0 
    open(file_path) do file
        prev_line = "no"
        for line in eachline(file)
            if line[1] == 'o'
                global o_tested += 1
            end
            if prev_line != "no"
                if line[1] == 'f' && prev_line[1] == 'f'
                    global f_tested += 1
                    @assert length(split(strip(line))[2:end]) == length(split(strip(prev_line))[2:end])
                end
            end
            prev_line = line
        end
    end
end
print("hehehehehehhe we're good")
```

# camera location help

```julia
function lazy_bounds(fname::String)
    shapes = []
    for obj_file in [fname]
        obj_path = joinpath(dirpath, obj_file)
        objects = RayTracing.parse_obj(
            obj_path,
            RayTracing.Translate(RayTracing.Pnt3(0,0,0)),
            false,
            false,
            nothing
        )
        for object in objects
            for mesh in object
                push!(shapes, mesh)
            end
        end
    end

    primitives = RayTracing.Primitive[]

    mat_gray = RayTracing.Matte(
        RayTracing.ConstantTexture(RayTracing.spectrum_from_float(.9, .9, .795)),
        RayTracing.ConstantTexture(0.0),
        nothing
    )

    for shape in shapes
        push!(primitives, RayTracing.Primitive(shape, mat_gray, nothing))
    end


    bvh = RayTracing.BVH(primitives)

    b = RayTracing.world_bounds(bvh)

    return b, (b.pMax + b.pMin) ./ 2.0
end
```

```julia
lazy_bounds("mesh_00113.obj") # pool surface
```
