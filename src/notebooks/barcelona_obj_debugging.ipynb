# let's assert that within an object in an obj, it is always either tris or quads..... no mixing please

```julia
file_path = "/Users/johnmyslinski/Documents/pbrt-v3-scenes/barcelona-pavilion/geometry/mesh_00001.obj"
```

```julia
f_tested = 0 
o_tested = 0 
open(file_path) do file
    prev_line = "no"
    for line in eachline(file)
        if line[1] == 'o'
            global o_tested += 1
        end
        if prev_line != "no"
            if line[1] == 'f' && prev_line[1] == 'f'
                global f_tested += 1
                @assert length(split(strip(line))[2:end]) == length(split(strip(prev_line))[2:end])
            end
        end
        prev_line = line
    end
end
```

```julia
f_tested + o_tested
```
