# let's assert that within an object in an obj, it is always either tris or quads..... no mixing please

```julia
objs = String[]
dirpath = "/Users/johnmyslinski/Documents/pbrt-v3-scenes/barcelona-pavilion/geometry/"
for (root, dir, files) in walkdir(dirpath)
    for file in files
        if endswith(file, ".obj")
            push!(objs, file)
        end
    end
end
```

```julia
for obj in objs
    file_path = dirpath * obj

    f_tested = 0 
    o_tested = 0 
    open(file_path) do file
        prev_line = "no"
        for line in eachline(file)
            if line[1] == 'o'
                global o_tested += 1
            end
            if prev_line != "no"
                if line[1] == 'f' && prev_line[1] == 'f'
                    global f_tested += 1
                    @assert length(split(strip(line))[2:end]) == length(split(strip(prev_line))[2:end])
                end
            end
            prev_line = line
        end
    end
end
print("hehehehehehhe we're good")
```

# dig into one of 'em

```julia
shapes = RayTracing.parse_obj(
    "/Users/johnmyslinski/Documents/pbrt-v3-scenes/barcelona-pavilion/geometry/mesh_00001.obj",
    RayTracing.Translate(RayTracing.Pnt3(0,0,0)),
    false,
    false,
    nothing
)
```

```julia
tri = shapes[1]
```

```julia
for each in tri
    if !(each.mesh.vertices isa Nothing)
        @assert length(each.mesh.vertices) == maximum(each.mesh.vertex_indices)
    end
    if !(each.mesh.normals isa Nothing)
        @assert length(each.mesh.normals) == maximum(each.mesh.normal_indices)
    end
    if !(each.mesh.uvs isa Nothing)
        @assert length(each.mesh.uvs) == maximum(each.mesh.uv_indices)
    end
end
```

# count shapes

```julia
shapes = []
for obj_file in objs
    obj_path = joinpath(dirpath, obj_file)
    objects = RayTracing.parse_obj(
        obj_path,
        RayTracing.Translate(RayTracing.Pnt3(0,0,0)),
        false,
        false,
        nothing
    )
    for object in objects
        for mesh in object
            push!(shapes, mesh)
        end
    end
end
```

# camera location help

```julia
shapes = []
for obj_file in ["mesh_00113.obj"]
    obj_path = joinpath(dirpath, obj_file)
    objects = RayTracing.parse_obj(
        obj_path,
        RayTracing.Translate(RayTracing.Pnt3(0,0,0)),
        false,
        false,
        nothing
    )
    for object in objects
        for mesh in object
            push!(shapes, mesh)
        end
    end
end
```

```julia
primitives = RayTracing.Primitive[]

mat_gray = RayTracing.Matte(
    RayTracing.ConstantTexture(RayTracing.spectrum_from_float(.9, .9, .795)),
    RayTracing.ConstantTexture(0.0),
    nothing
)

for shape in shapes
    push!(primitives, RayTracing.Primitive(shape, mat_gray, nothing))
end


bvh = RayTracing.BVH(primitives)
```

```julia
b = RayTracing.world_bounds(bvh)
(b.pMax - b.pMin) ./ 2.0
```
