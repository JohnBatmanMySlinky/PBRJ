```julia
include("../RayTracing.jl")
```

```julia
test_fp = "/home/jmyslinski/random_stuff/PBRJ/src/ld-9-power.exr"
baseline_fp = "/home/jmyslinski/random_stuff/PBRJ/src/ld-9-uniform.exr"
```

```julia
test, test_L, test_W = RayTracing.read_image(test_fp, RayTracing.spectrum_from_float(1.0))
baseline, baseline_L, baseline_W = RayTracing.read_image(baseline_fp, RayTracing.spectrum_from_float(1.0))
```

```julia
@assert test_L == baseline_L
@assert test_W == baseline_W
```

```julia
tot = RayTracing.spectrum_from_float(0.0)
i = 0
for y in 1:test_L
    for x in 1:test_W
        i += 1
        tot = tot + (test[i] - baseline[i]) .^ 2
    end
end
RayTracing.sqrt.(tot ./ (test_L * test_W))
```
