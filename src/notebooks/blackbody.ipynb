```julia
function blackbody(lambda::Vector{Float64}, T::Float64)
    LL = zeros(Float64, n)
    if T < 0.0
        return LL
    end
    c = 299792458.0
    h = 6.62606957e-34
    kb = 1.3806488e-23
    for i in 1:n
        # Compute emitted radiance for blackbody at wavelength _lambda[i]_
        l = lambda[i] * 1e-9
        LL[i] = (2 * h * c^2) / (l^5 * (exp((h * c) / (l * kb *T)) - 1))
    end
    return LL
end
```

```julia
v = blackbody(RayTracing.CIE_lambda, length(RayTracing.CIE_lambda), 1_012.0)
```
